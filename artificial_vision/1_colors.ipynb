```python
import numpy as np
import matplotlib.pyplot as plt

def show(image, title = "Image"):
    plt.imshow(image, interpolation="nearest", cmap = 'gray')
    plt.axis("off")
    plt.show()
    
def shows(images, titles):
    n = len(images)
    fig, axs = plt.subplots(1, n)
    fig.set_size_inches(10, 3.5)
    
    for i in range(n):
        axs[i].set_title(titles[i])
        axs[i].axis('off')
        axs[i].imshow(images[i], cmap = 'gray')
        
    plt.show()
```

```python
# Load the cat for the RGB channel comparison.
cat = plt.imread("images_notebook/cat.png")
rgb = cat[:, :, :3] 

show(rgb)
```

```python
images = [cat[:, :, i] for i in range(3)]
titles = ["Red", "Green", "Blue"]

shows(images, titles)
```

```python
new_cat = cat.copy()
r, g, b = cat[:, :, 0], cat[:, :, 1], cat[:, :, 2]
```

```python
pupils =  (r < 0.2) & (b > 0.6)
new_cat[pupils, 1] = cat[pupils, 2]
new_cat[pupils, 2] = 0.2

shows([pupils, new_cat, cat], ["Pupils Mask", "New Cat", "Old Cat"])
```

```python
light_fur = (r > 0.1) & (b > 0.1) & (b < 0.6) & (g < 0.9)
new_cat[light_fur, 0:3] = 0.95

shows([light_fur, new_cat, cat], ["Light Fur", "New Cat", "Old Cat"])
```
